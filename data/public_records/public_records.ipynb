Création d'une fonction cleaning que j:'appelerais avec des paramètres différents pour chaque table (selon le besoin):
- pour remplir les valeurs manquantes: fillna_map
- convertir les colonnes contenant une date (si le format n'est pas bon): date_cols
- conversion du type de colonne: dtype_map
- vérifier le respect de format: format_checks
- supprimer les doublons selon le primary key (si fourni, on conserve qu'une ligne par key, sinon, on supprime les lignes ENTIEREMENT identiques)

Ajout d'une fonction à part qui traitera la lecture, le type et format des colonnes date

```python
def mixed_dates(series: pd.Series) -> pd.Series:
    """Essaie plusieurs formats de date dans l'ordre, jusqu'à ce que ça marche."""
    formats_to_try = [
        "%Y-%m-%d",                # dates simples
        "%d/%m/%Y",
        "%Y/%m/%d",
        "%Y-%m-%dT%H:%M:%S",       # datetimes avec heure
        "%Y/%m/%d %H:%M:%S",
        "%d-%b-%Y %H:%M:%S",       # ex: 08-Jun-2022 10:00:00
    ]
    parsed = pd.to_datetime(series, format=formats_to_try[0], errors="coerce")

    for fmt in formats_to_try[1:]:
        still_missing = parsed.isna() & series.notna()
        if not still_missing.any():
            break
        parsed.loc[still_missing] = pd.to_datetime(
            series.loc[still_missing], format=fmt, errors="coerce"
        )
    return parsed
```

Fonction fill_from_duplicate_entity permet de récupérer une information manquante à partir d'une entité dupliquée d'un même fichier.
On utilise cette fonction pour le nettoyage du fichier public_appointments où on veut récupérer l'année de naissance pour un "office_holder_name" qui existe déjà dans le fichier avec l'info nécessaire.
HYPOTHÈSE: on suppose qu'il n'y a pas d'homonyme sur les prénoms, qu'il s'agit de personnes différentes.
- lookup construit une table du type {nom : année de naissance} à partir des infos présentes.
- avec .map() pour récupérer ces infos et combler les vides des colonnes en question

---> inférence de données

```python
def fill_from_duplicate_entity(df: pd.DataFrame, entity_col: str, target_col: str) -> pd.DataFrame:
    df = df.copy()
    lookup = df.dropna(subset=[target_col]).drop_duplicates(subset=[entity_col])
    lookup = lookup.set_index(entity_col)[target_col]

    mask = df[target_col].isna()
    df.loc[mask, target_col] = df.loc[mask, entity_col].map(lookup)
    return df
```

pour la suppression de ligne, si pk_col est défini, on ne garde qu'une ligne par valeur unique de la PK (deux lignes avec le même registration_number = un doublon, même si le reste diffère) ; si pk_col n'est pas défini, subset=None fait comparer toutes les colonnes (doublon strict, comme df.duplicated() plus haut).

```python
def clean_pipeline(df: pd.DataFrame, name: str, pk_col: str = None,
                    date_cols: list = None, fillna_map: dict = None, format_checks: dict = None) -> pd.DataFrame:
    df = df.copy()

    if fillna_map:
        df = df.fillna(fillna_map)

    if date_cols:
        for c in date_cols:
            df[c] = mixed_dates(df[c])
            # coerce ici permet de transformer les dates invalides en NaT

    before = len(df)
    df = df.drop_duplicates(subset=[pk_col] if pk_col else None)
    if len(df) < before:
        print(f"[{name}] {before - len(df)} doublons supprimés")

    if pk_col:
        n_null = df[pk_col].isna().sum() # on vérifie pk_col == 0 pour être sur quil n'y pas de manquant dans les clés primaires
        n_dup = df[pk_col].duplicated().sum() # pareil, si != 0, il y a des doublons
        status = "OK" if (n_null == 0 and n_dup == 0) else "A VERIFIER"
        print(f"[{name}] PK '{pk_col}' -> {n_null} manquants, {n_dup} doublons [{status}]")

    if format_checks:
        for col, pattern in format_checks.items():
            non_conformes = df[~df[col].astype(str).str.match(pattern, na=False)]
            print(f"[{name}] '{col}' non conformes au pattern : {len(non_conformes)}")

    return df
```

Création du dictionnaire de configuration CONFIG, avec comme clés externes les noms de fichier, et valeur des dictionnaires des paramètres à passer pour executer la fonction clean_pipeline sur chaque fichier

```python
CONFIG = {
    "registered_entities": {
        "pk_col": "registration_number",
        "date_cols": ["registration_date", "last_filing_date"],
        "format_checks": {
            "registration_number": r'^REG-[A-Z]{3}-\d{5}$',
    },
    },
    "licensed_facilities": {
        "pk_col": "facility_license_number",
        "date_cols": ["license_start", "license_end"],
        "format_checks": {
            "facility_license_number": r'^FAC-[A-Z]{3}-\d{3}$',
            "operating_organization_registration": r'^REG-[A-Z]{3}-\d{5}$',
    },
    },
    "licensed_professionals": {
        "pk_col": "license_number",
        "date_cols": ["license_issue_date", "license_expiry_date"],
        "format_checks": {
            "license_number": r'^LIC-[A-Z]{3}-[A-Z]{3}-\d{4}$',
            },
    },
    "officer_filings": {
        "pk_col": "filing_id",
        "date_cols": ["appointment_date", "cessation_date", "filing_date"],
        "format_checks": {
            "filing_id": r'^FLG-\d{6}$',
            "registration_number": r'^REG-[A-Z]{3}-\d{5}$',
            "officer_local_ref": r'^OFR-\d{5}$',
    },
    },
    "regulatory_case_events": {
        "pk_col": "regulatory_event_id",
        "date_cols": ["event_time"],
        "format_checks": {
            "regulatory_event_id": r'^RGE-\d{6}$',
            "case_number": r'^REG-\d{4}-\d{4}$',
    },
    },
    "site_access_case_events": {
        "pk_col": "access_event_id",
        "date_cols": ["event_time"],
        "format_checks": {
            "access_event_id": r'^ACE-\d{6}$',
            "access_case_number": r'^ACC-\d{4}-\d{4}$',
    },
    },
    "public_appointments": {
        "pk_col": "appointment_record_id",
        "date_cols": ["appointment_start", "appointment_end"],
        "format_checks": {
            "appointment_record_id": r'^APT-\d{6}$',
            "source_notice_reference": r'^NOTICE-[A-Z]{3}-\d{4}-\d{4}$',
    },
    },
    "institutional_memos": {
        "pk_col": "document_id",
        "date_cols": ["document_date"],
        "format_checks": {
            "document_id": r'^MEM-\d{6}$',
    },
    },
    "public_and_investigative_documents": {
        "pk_col": "document_id",
        "date_cols": ["document_date"],
        "format_checks": {
            "document_id": r'^PUB-\d{6}$',
    },
    },
}
```

## Corrections spécifiques par fichier
Fichiers nécessitants un nettoyage spécifique en plus du général

##### licensed_professionals

```python
df = raw_dfs["licensed_professionals"].copy()
# corrige les espaces multiples dans la colonne specialty
df["specialty"] = df["specialty"].str.strip().str.replace(r"\s+", " ", regex=True)
raw_dfs["licensed_professionals"] = df
```

##### officer_filings

```python
df = raw_dfs["officer_filings"].copy()
df["full_name"] = df["full_name"].str.strip().str.replace(r"\s+", " ", regex=True)
raw_dfs["officer_filings"] = df
```

##### public_appointments

```python
df = raw_dfs["public_appointments"].copy()
# corrige les espaces multiples dans la colonne institution_name
df["institution_name"] = df["institution_name"].str.strip().str.replace(r"\s+", " ", regex=True)
# appel de la fonction fill_from_duplicate_entity pour remplir les valeurs manquantes de recorded_birth_year
df = fill_from_duplicate_entity(df, entity_col="office_holder_name", target_col="recorded_birth_year")
raw_dfs["public_appointments"] = df
```

##### institutional_memos

```python
df = raw_dfs["institutional_memos"].copy()
# corrige les espaces multiples dans la colonne subject_line
df["subject_line"] = df["subject_line"].str.strip().str.replace(r"\s+", " ", regex=True)
raw_dfs["institutional_memos"] = df
```

## Application et export

```python
clean_dfs = {
    name: clean_pipeline(df, name=name, **CONFIG[name])
    for name, df in raw_dfs.items()
}
```

## Export des CSV nettoyés

```python
for name, df in clean_dfs.items():
    df.to_csv(f"{name}_cleaned.csv", index=False)

print(f"{len(clean_dfs)} fichiers exportés dans {Path('.').resolve()}")
```

## Contrôle final — statistiques sur les CSV nettoyés

Construit un tableau récapitulatif : une ligne par fichier, avec les stats clés.
`label` permet de distinguer 'brut' vs 'nettoyé' une fois les deux tableaux combinés.

```python
def summarize_files(dfs: dict, label: str) -> pd.DataFrame:
    rows = []
    for name, df in dfs.items():
        rows.append({
            "fichier": name,
            "etat": label,
            "n_lignes": df.shape[0],
            "n_colonnes": df.shape[1],
            "pct_manquant_moyen": round(df.isna().mean().mean() * 100, 2),
            "colonne_plus_vide": df.isna().mean().idxmax(),
            "pct_max_manquant": round(df.isna().mean().max() * 100, 2),
            "n_doublons": df.duplicated().sum(),
        })
    return pd.DataFrame(rows).sort_values("fichier").reset_index(drop=True)

summary_raw = summarize_files(raw_dfs, "brut")
summary_clean = summarize_files(clean_dfs, "nettoyé")

summary_comparison = pd.concat([summary_raw, summary_clean], ignore_index=True)
summary_comparison = summary_comparison.sort_values(["fichier", "etat"]).reset_index(drop=True)
display(summary_comparison)
```
